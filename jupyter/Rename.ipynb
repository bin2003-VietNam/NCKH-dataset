```python
import os



folder_path = "C:/Users/nguye/jupyter/dataset_sorted/21"
prefix = "0021_"  # phần đầu cố định
# Lấy danh sách file và sắp xếp
files = sorted(os.listdir(folder_path))



# Bước 1: đổi tạm để tránh trùng tên
for filename in files:
    if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
        temp_name = "temp_" + filename
        os.rename(os.path.join(folder_path, filename),
                  os.path.join(folder_path, temp_name))

# Bước 2: rename thành dạng 1000_0000
temp_files = sorted(os.listdir(folder_path))
for i, filename in enumerate(temp_files):
    if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
        ext = os.path.splitext(filename)[1]
        new_name = f"{prefix}{i:04d}{ext}"  # 1000_0000.jpg
        os.rename(os.path.join(folder_path, filename),
                  os.path.join(folder_path, new_name))

print("✅ Rename done! Example:", new_name)
```
